```python
import pandas as pd

fake = pd.read_csv("../data/Fake.csv")
true = pd.read_csv("../data/True.csv")

print(fake.head())
print(true.head())
```

```python
print(fake.columns)

print(fake.shape)

print(fake.info())
```

```python
print(fake.isnull().sum())
```

```python
fake["length"] = fake["text"].apply(len)

print(fake["length"].mean())
```

```python
import matplotlib.pyplot as plt

fake["length"].hist()

plt.title("Fake News Article Length")

plt.show()
```

```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt

text = " ".join(fake["text"].astype(str))

wordcloud = WordCloud(width=800, height=400).generate(text)

plt.figure(figsize=(10,5))
plt.imshow(wordcloud)
plt.axis("off")
plt.show()
```
